```python
from ollama import chat

system_prompt = {"role": "system", "content": "请使用简体中文回答所有问题"}


def do_chat(
    prompt,
    msg_history: list = None,
    temperature=0,
    max_tokens=1024,
    num_predict=500,
):
    if msg_history is None:
        msg_history = []
    msgs = [system_prompt]
    msgs.extend(msg_history)
    user_input = {"role": "user", "content": prompt}
    msgs.append(user_input)
    msg_history.append(user_input)
    stream = chat(
        # model="llama3:8b",
        model="qwen2.5:7b",
        messages=msgs,
        stream=True,
        options={
            "temperature": temperature,
            "max_tokens": max_tokens,
            "num_predict": num_predict,
        },
    )
    resp = []
    # 逐块打印响应内容
    for chunk in stream:
        msg = chunk["message"]["content"]
        print(msg, end="", flush=True)
        resp.append(msg)
    msg_history.append({"role": "assistant", "content": "".join(resp)})
    return msg_history
```

```python

import pandas as pd
import numpy as np
from openai import OpenAI





def load_dataset(source: str) -> pd.core.frame.DataFrame:
    # Load the video session index
    pd_vectors = pd.read_json(source)
    return pd_vectors.drop(columns=["text"], errors="ignore").fillna("")


def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))


def get_videos(
    query: str, dataset: pd.core.frame.DataFrame, rows: int
) -> pd.core.frame.DataFrame:
    # create a copy of the dataset
    video_vectors = dataset.copy()

    # get the embeddings for the query
    query_embeddings = (
        client.embeddings.create(input=query, model=model).data[0].embedding
    )

    # create a new column with the calculated similarity for each row
    video_vectors["similarity"] = video_vectors["nomic_embed_text"].apply(
        lambda x: cosine_similarity(np.array(query_embeddings), np.array(x))
    )

    # filter the videos by similarity
    mask = video_vectors["similarity"] >= SIMILARITIES_RESULTS_THRESHOLD
    video_vectors = video_vectors[mask].copy()

    # sort the videos by similarity
    video_vectors = video_vectors.sort_values(by="similarity", ascending=False).head(
        rows
    )

    # return the top rows
    return video_vectors.head(rows)


def display_results(videos: pd.core.frame.DataFrame, query: str):
    def _gen_yt_url(video_id: str, seconds: int) -> str:
        """convert time in format 00:00:00 to seconds"""
        return f"https://youtu.be/{video_id}?t={seconds}"

    print(f"\nVideos similar to '{query}':")
    for _, row in videos.iterrows():
        youtube_url = _gen_yt_url(row["videoId"], row["seconds"])
        print(f" - {row['title']}")
        print(f"   Summary: {' '.join(row['summary'].split()[:15])}...")
        print(f"   YouTube: {youtube_url}")
        print(f"   Similarity: {row['similarity']}")
        print(f"   Speakers: {row['speaker']}")


# DATASET_NAME = "08-building-search-applications/scripts/embedding_index_3m.json"
DATASET_NAME = "08-building-search-applications/scripts/embedding_index_3m_nomic.json"
SIMILARITIES_RESULTS_THRESHOLD = 0.8
model = "nomic-embed-text:latest"


client = OpenAI(
    api_key="ollama",
    base_url="http://127.0.0.1:11434/v1",
)

pd_vectors = load_dataset(DATASET_NAME)

query = "What is Azure Machine Learning?"

videos = get_videos(query, pd_vectors, 5)
display_results(videos, query)
```
